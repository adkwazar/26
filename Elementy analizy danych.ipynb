<h4> TF-IDF

Technika zamiany tekstu na wektory liczbowe

```python
documents = ["new york times", "new york post", "los angeles times"]
documents = [elem.split() for elem in documents]
print(documents)
N = len(documents)
print(N)
```

```python
import binf2 as b
import pandas as pd

df = b.TFIDF(documents)
print(df)
```

<h4> Standaryzacja, redukcja wymiaru (PCA) i wizualizacja

```python
#tutaj uzylem pandas, mozna tez za pomocą polars, import polars as pl

df = pd.DataFrame({
    "A": [12, 15, 18, 21, 24, 27, 30, 33, 36, 39],
    "B": [4, 7, 9, 12, 15, 18, 20, 23, 26, 29],
    "C": [22, 19, 25, 28, 24, 31, 35, 32, 38, 40],
    "D": [8, 11, 14, 17, 13, 20, 23, 26, 22, 29],
    "E": [30, 28, 35, 32, 40, 38, 45, 42, 50, 48],
    "F": [5, 10, 15, 12, 20, 25, 22, 30, 35, 40]
})

print(df)
```

```python
from sklearn.preprocessing import StandardScaler 
sc = StandardScaler() #standaryzacja danych
Xs = sc.fit_transform(df)

from sklearn.decomposition import PCA
pca = PCA(n_components=2) #redukcja 6 kolumn wejsciowych na nowe 2 cechy
pca.fit(Xs)
X_pca = pca.transform(Xs)

import matplotlib.pyplot as plt 
plt.scatter(X_pca[:, 0], X_pca[:, 1], alpha=1, c = ["red"]*3+["blue"]*7) #wizulizacja
plt.show()
```
